# Data Wrangling Process

```python
dataset = dataset.iloc[:, 1:9]
dataset.head()
```

```python
# Convert date and time to datetime format
dataset['datetime'] = pd.to_datetime(dataset['date'] + ' ' + dataset['time'])
dataset['day_of_week'] = dataset['datetime'].dt.dayofweek
```

```python
dataset.head()
```

```python
dataset.drop(columns=['date', 'time', 'venue', 'description', 'datetime'], inplace=True)
```

```python
# Encode categorical data
label_enc = LabelEncoder()
dataset['type_encoded'] = label_enc.fit_transform(dataset['type'])
scaler = StandardScaler()
dataset[['price', 'day_of_week']] = scaler.fit_transform(dataset[['price', 'day_of_week']])
```

```python
# Feature set for clustering
features = dataset[['price', 'day_of_week', 'type_encoded']]
```

```python
# KMeans Clustering
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
dataset['KMeans_Cluster'] = kmeans.fit_predict(features)

# DBSCAN Clustering
dbscan = DBSCAN(eps=1.5, min_samples=2)
dataset['DBSCAN_Cluster'] = dbscan.fit_predict(features)

# Gaussian Mixture Model Clustering
gmm = GaussianMixture(n_components=3, random_state=42)
dataset['GMM_Cluster'] = gmm.fit_predict(features)
```

```python
dataset.head()
```

```python
# Save models as .pkl files
with open('./trained_models/kmeans_model.joblib', 'wb') as f:
    joblib.dump(kmeans, f)

with open('./trained_models/dbscan_model.joblib', 'wb') as f:
    joblib.dump(dbscan, f)

with open('./trained_models/gmm_model.joblib', 'wb') as f:
    joblib.dump(gmm, f)

with open('./trained_models/scaler.joblib', 'wb') as f:
    joblib.dump(scaler, f)

with open('./trained_models/label_encoder.joblib', 'wb') as f:
    joblib.dump(label_enc, f)
```

```python
data ={
    "title": "Tech Expo 2025",
    "type": "Workshop",
    "price": 1200,
    "city": "Chennai",
    "day_of_week": 3
}
```

```python
# Convert JSON to DataFrame
data = pd.DataFrame([data])
print(data)

# Extract and preprocess input features
price = float(data.loc[0, 'price'])  # Accessing the first row
day_of_week = int(data.loc[0, 'day_of_week'])
event_type = data.loc[0, 'type']
```

```python
# Encode event type
type_encoded = label_enc.transform([[event_type]])[0]  # Ensure it's a list before transforming
```

```python
print(type_encoded)
```

```python
# Convert features to a DataFrame and scale
features = pd.DataFrame([[price, day_of_week]], columns=['price', 'day_of_week'])
print(features)
```

```python
features_scaled = scaler.transform(features)  # Transform correctly
print(features_scaled)
```

```python
# Convert scaled features back into a DataFrame with correct column names
features_scaled_df = pd.DataFrame(features_scaled, columns=['price', 'day_of_week'])

# Add type_encoded back to the scaled DataFrame
features_scaled_df["type_encoded"] = type_encoded
```

```python
print(features_scaled_df)
```

```python
 # Get predictions
kmeans_cluster = int(kmeans.predict(features_scaled_df)[0])
gmm_cluster = int(gmm.predict(features_scaled_df)[0])

# Handle DBSCAN prediction
dbscan_cluster = -1  # Default to noise
```

```python
dbscan_cluster = int(dbscan.fit_predict(features_scaled_df)[0])
```

```python
print(dbscan_cluster)
```

```python
# Get recommended events, ignoring DBSCAN cluster if it's -1
recommended_events = dataset[
    (dataset["KMeans_Cluster"] == kmeans_cluster) |
    (dataset["GMM_Cluster"] == gmm_cluster) |
    ((dataset["DBSCAN_Cluster"] == dbscan_cluster) & (dbscan_cluster != -1))
    ][["title", "type", "city"]].to_dict(orient="records")
```
